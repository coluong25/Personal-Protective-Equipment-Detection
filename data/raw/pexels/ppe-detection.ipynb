# SH17 Dataset

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
class_name = {
11: 'hand', 
0: 'person',
12: 'head',
3: 'face',
1: 'ear',
7: 'tool',
14: 'shoes',
9: 'gloves',
8: 'glasses',
10: 'helmet',
6: 'foot',
5: 'face-mask-medical',
16: 'safety-vest',
2: 'ear-mufs',
15: 'safety-suit',
13: 'medical-suit',
4: 'faceguard'
}

```

```python
import os
dir_list = os.listdir('/kaggle/input/sh17-dataset-for-ppe-detection/labels')

must_have_class = [9, 10, 15, 16]
new_file = open("usable_image.txt", "w")
for filename in dir_list:
    with open('/kaggle/input/sh17-dataset-for-ppe-detection/labels/' + filename) as f:
        f = f.readlines()
        classes_in_img = []
        for line in f:
            line = line.rstrip()
            line = line.split(' ')
            classes_in_img.append(int(line[0]))
        is_duplicate = bool(set(must_have_class).intersection(classes_in_img))
        if is_duplicate:
            new_file.write(filename.split('.')[0]+'\n')
new_file.close()
```

```python
with open('/kaggle/working/usable_image.txt') as f:
    f = f.readlines()
    print(len(f))
```

```python
if not os.path.exists('data/labels'):
    os.makedirs('data/labels')
if not os.path.exists('data/images'):
    os.makedirs('data/images')
```

```python
chosen_class = [0, 9, 10, 15, 16]
new_label_class = [0, 2, 1, 4, 4]
with open('/kaggle/working/usable_image.txt') as f1:
    f1 = f1.readlines()
    for line1 in f1:
        line1 = line1.rstrip()
        new_txt = open(f"/kaggle/working/data/labels/{line1}.txt", "w")
        try:
            img = cv2.imread('/kaggle/input/sh17-dataset-for-ppe-detection/images/' + line1 + '.jpeg')
            cv2.imwrite('/kaggle/working/data/images/'+ line1 + '.jpg', img)
        except:
            continue
        img_y, img_x = img.shape[:2]
        with open('/kaggle/input/sh17-dataset-for-ppe-detection/labels/' + line1 + '.txt') as f:
            f = f.readlines()
            for line in f:
                cur_class = line.rstrip().split(' ')[0]
                if int(cur_class) in chosen_class:
                    new_line = line.replace(str(cur_class), str(new_label_class[chosen_class.index(int(cur_class))]), 1)
                    new_txt.write(new_line)
        new_txt.close()
           
```
